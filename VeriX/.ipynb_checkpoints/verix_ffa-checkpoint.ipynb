```python
from VeriX import *
from enum_axps import AXPEngine
import time
import sklearn
from sklearn.preprocessing import StandardScaler
```

```python
dataset = sklearn.datasets.load_breast_cancer()
X_train, y_train = dataset.data[:400], dataset.target[:400]
X_test, y_test = dataset.data[400:], dataset.target[400:]

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
X_train = X_train.astype(np.float32)
X_test = X_test.astype(np.float32)
```

```python
v = VeriX(
    datatype="tabular",
    instance=X_test[0],
    model_path="models/breast_cancer.onnx",
    plot_original=False,
)

v.epsilon = 0.05
```

```python
engine = AXPEngine(v)
axps = engine.enumerate_axps()

print("AXPs found:")
for a in axps:
    print(sorted(list(a)))
```

```python
from anytime_axp import AnytimeAXP
```

```python
v = VeriX(
    datatype="tabular",
    instance=X_test[0],
    model_path="models/breast_cancer.onnx",
    plot_original=False,
)

v.epsilon = 0.2
```

```python
engine = AnytimeAXP(v)

axps = engine.run(iterations=100)
```

```python
v = VeriX(
    datatype="tabular",
    instance=X_test[0],
    model_path="models/breast_cancer.onnx",
    plot_original=False,
)

v.traversal_order(traverse="heuristic")

v.get_explanation(
    epsilon=0.5,
    plot_counterfactual=False,
    plot_explanation=False
)
```

```python
time1 = time.time()
verix = VeriX(
    datatype="tabular",
    instance=X_test[0],
    model_path="models/verix_synthetic.onnx",
    plot_original=False
)

verix.traversal_order(traverse="heuristic")

verix.get_explanation(
    epsilon=0.15,
    plot_counterfactual=False,
    plot_explanation=False
)
time2 = time.time()
print(f"VeriX_FFA time: {time2 - time1:.4f} seconds")
```

```python
# remove useless and incorrect information
print(f"Dataset size before:\t{X.shape[0]}")
X = X.drop(X[X['MARRIAGE']==0].index)
X = X.drop(X[X['EDUCATION']==0].index)
X = X.drop(X[X['EDUCATION']==5].index)
X = X.drop(X[X['EDUCATION']==6].index)
print(f"Dataset size after:\t{X.shape[0]}")
```

```python
pay_features = ['PAY_1', 'PAY_2', 'PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']
for p in pay_features:
    X.loc[X[p]<0, p] = -1
    X.loc[X[p]>=0, p] = X.loc[X[p]>=0, p] + 1
    X[p] = X[p].astype('int64')
```

```python
X['GRAD_SCHOOL'] = (X['EDUCATION'] == 1).astype('category')
X['UNIVERSITY'] = (X['EDUCATION'] == 2).astype('category')
X['HIGH_SCHOOL'] = (X['EDUCATION'] == 3).astype('category')
X.drop('EDUCATION', axis=1, inplace=True)

X['MALE'] = (X['SEX'] == 1).astype('category')
X.drop('SEX', axis=1, inplace=True)

X['MARRIED'] = (X['MARRIAGE'] == 1).astype('category')
X.drop('MARRIAGE', axis=1, inplace=True)

X.head()
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

def scale_features(X_train, X_test, columns, scaler):
    X_train_scaled = X_train.copy()
    X_test_scaled = X_test.copy()

    for col in columns:
        X_train_scaled[col] = scaler.fit_transform(X_train[col].values.reshape(-1, 1))
        X_test_scaled[col] = scaler.transform(X_test[col].values.reshape(-1, 1))

    return X_train_scaled, X_test_scaled

# Columns to be scaled/standardized
# columns = ['LIMIT_BAL', 'AGE'] + [f'PAY_{i}' for i in range(1, 7)] + [f'BILL_AMT{i}' for i in range(1, 7)] + [f'PAY_AMT{i}' for i in range(1, 7)]

columns = ['LIMIT_BAL', 'AGE'] + [f'BILL_AMT{i}' for i in range(1, 7)] + [f'PAY_AMT{i}' for i in range(1, 7)]
#
# # MinMax scaling (normalized)
# scaler_minmax = MinMaxScaler()
# X_train_norm, X_test_norm = scale_features(X_train_raw, X_test_raw, columns, scaler_minmax)

# Standard scaling
scaler_standard = StandardScaler()
X_train_std, X_test_std = scale_features(X_train_raw, X_test_raw, columns, scaler_standard)

# # Rename columns to indicate scaling type
# X_train_norm.columns = [col + '_norm' for col in X_train_norm.columns]
# X_test_norm.columns = [col + '_norm' for col in X_test_norm.columns]

X_train_std.columns = [col + '_std' for col in X_train_std.columns]
X_test_std.columns = [col + '_std' for col in X_test_std.columns]

# Combine the datasets
# X_train_combined = pd.concat([X_train_norm, X_train_std], axis=1)
# X_test_combined = pd.concat([X_test_norm, X_test_std], axis=1)
X_train_combined = X_train_std
X_test_combined = X_test_std
```
